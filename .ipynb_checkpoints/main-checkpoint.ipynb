#  LendingClub Credit Risk Analysis & Stress Testing

## Projekt áttekintése
Ez a projekt a LendingClub P2P hitelezési platform nyilvános adatain keresztül vizsgálja a hitelkockázati trendeket és a portfólió ellenálló képességét. A cél egy átfogó elemzés készítése, amely feltárja a bedőlés (default) valószínűségét befolyásoló kulcstényezőket, majd egy stresszteszt keretében szimulálja, hogyan reagálna a portfólió egy kedvezőtlen makrogazdasági környezetre.

**Főbb lépések:**
1.  Exploratory Data Analysis (EDA): A kockázati driverek (pl. jövedelem, eladósodottság, hitelminősítés) és a default ráta kapcsolatának feltárása.
2.  Kockázati modellezés: A portfólió minőségének vizsgálata idősoros és szegmentált bontásban.
3.  Stresszteszt (Scenario Analysis): Makrogazdasági sokkok (jövedelemkiesés, munkanélküliség) szimulálása és hatásuk becslése a portfólió várható veszteségére (PD Uplift).

### Default definíció és Időbeli Elemzés

Az elemzés során a defaultot konzervatív módon állapítom meg: csak a *Fully Paid* (teljesen visszafizetett) hitelek nem minősülnek defaultnak. A futó (Current) hitelek kizárásra kerülnek, mivel ezek végső kimenetele még nem ismert.

A portfólió minőségét **kihelyezési év szerinti bontásban** vizsgálom. Bár ez egyszerűbb, mint egy teljes cohort (vintage) elemzés, arra tökéletesen alkalmas, hogy megmutassa, hogyan változott a kockázati szint a különböző években (pl. válság utáni időszak vs. konjunktúra).

Forrás a kockázati mutatókhoz:
[Corporate Finance Institute - Expected Loss](https://corporatefinanceinstitute.com/resources/career-map/sell-side/risk-management/expected-loss-definition-calculation-importance/)

```python
DEFAULT = ["Charged Off", "Default"]
NON_DEFAULT = ["Fully Paid"]

df = df[df["loan_status"].isin(DEFAULT + NON_DEFAULT)].copy()
df["defaultFlag"] = df["loan_status"].isin(DEFAULT).astype(int)


df['issue_d'] = pd.to_datetime(df['issue_d'], format="%b-%Y")
df["issue_year"] = df["issue_d"].dt.year


df.groupby("issue_year").size()

annual_pd = (
    df.groupby("issue_year")['defaultFlag'].mean()
    .reset_index(name="default_rate")
)
```

```python
plt.figure()

plt.plot(annual_pd['issue_year'], annual_pd['default_rate'], marker="o")
plt.xlabel("Kihelyezés éve")
plt.ylabel("Default arány")
plt.title("Default arány alakulása kihelyezési év szerint")
plt.grid(True)
plt.show()
```

#### Kockázati hajtóerők vizsgálata
Most megvizsgáljuk, hogy mely hitel és ügyféljellemzők mentén tér el érdemben a default kockázat

A cél annak megértése, hogy melyik tényezők járulnak hozzá a portfólió kockázatához

[Credit risk alapjairól a szekcióhoz itt olvastam](https://www.investopedia.com/terms/c/creditrisk.asp)

```python
# hitelminősítés szerinti bontás
grade_pd = (
    df.groupby("grade")["defaultFlag"]
    .mean()
    .sort_values(ascending=False)
)
grade_pd
```

```python
# DTI -> Eladósottsági mutató
df["dti_bucket"] = pd.qcut(df["dti"], 10, duplicates="drop")

dti_pd = (
    df.groupby("dti_bucket", observed=False)["defaultFlag"].mean()
)
dti_pd
```

```python
df["inc_bucket"] = pd.qcut(df['annual_inc'], 10, duplicates="drop")

df.groupby("inc_bucket", observed=False)["defaultFlag"].mean()
```

```python
fig, ax = plt.subplots(1,3, figsize=(18,5))

ax[0].bar(grade_pd.index, grade_pd.values)
ax[0].set_title("Default Arány Minősítés (Grade) szerint")
ax[0].set_xlabel("Grade")
ax[0].set_ylabel("Default Rate")
ax[0].set_ylim(0, 0.6) 
ax[0].grid(axis='y', linestyle='--', alpha=0.7)




ax[1].bar(dti_pd.index.astype(str), dti_pd.values, color="salmon")
ax[1].set_title("Default Arány DTI sávok szerint")
ax[1].set_xlabel("DTI Bucket")
ax[1].tick_params(axis='x', rotation=45)
ax[1].grid(axis='y', linestyle='--', alpha=0.7)

inc_pd_plot = df.groupby("inc_bucket", observed=False)["defaultFlag"].mean()

ax[2].bar(inc_pd_plot.index.astype(str), inc_pd_plot.values, color="lightgreen")
ax[2].set_title("Default Arány Jövedelem szerint")
ax[2].set_xlabel("Income Bucket")
ax[2].tick_params(axis='x', rotation=45)
ax[2].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### Tanulság a hajtóerőkről
Bebizonyosodott, amit logikusan is gondolni lehetett
- A hitelminősítés erős megkülönböztető erejű kockázati tényező. Az alacsonyabb grade-ekhez lényegesen magasabb default arány társul, ami összhangban van a hitelminősítési logikával.
- [DTI](https://www.investopedia.com/terms/d/dti.asp) növekedésével a default kockázat emelkedik, ami alátámasztja, hogy a törlesztési teher fontos kockázati hajtóerő.
- Az alacsonyabb jövedelmi sávokhoz magasabb default arány társul, ami a visszafizetési kapacitás korlátaira utal.

#### Stressz teszt 
Ebben a szekcióban a portfólió hitelkockázatának stresszérzékenységét vizsgáljuk több forgatókönyv mentén.
A stresszteszt célja nem egy „tökéletes” előrejelző modell, hanem a portfólió kockázati profiljának és érzékenységének átlátható bemutatása egy kedvezőtlenebb környezetben.

A forgatókönyvek több csatornán keresztül hatnak: (i) jövedelemcsökkenés, (ii) foglalkoztatási sokk (proxy), (iii) törlesztési teher növekedése (proxy). A megközelítés összhangban van a stressztesztelési gyakorlatban alkalmazott „scenario → risk driver → portfolio impact” logikával.

```python
SCENARIOS = {
    "baseline": {
        "income_shock_mu": 0.00,  # átlagos jövedelemváltozás
        "income_shock_sigma": 0.02,
        "unemp_prob": 0.00,       # kirúgás valószínűsége
        "payment_shock": 0.00,    # törlesztési teher
        "pd_uplift_target": 1.00  # portfólió PD szorzó cél
    },
    "adverse": {
        "income_shock_mu": -0.10,
        "income_shock_sigma": 0.07,
        "unemp_prob": 0.05,
        "payment_shock": 0.05,
        "pd_uplift_target": 1.50
    },
    "severe": {
        "income_shock_mu": -0.20,
        "income_shock_sigma": 0.10,
        "unemp_prob": 0.10,
        "payment_shock": 0.10,
        "pd_uplift_target": 2.20
    }
}
```

```python
df_stress = df.copy()

df_stress['annual_inc'] = df_stress['annual_inc'].fillna(df_stress['annual_inc'].median())
df_stress['installment'] = df_stress['installment'].fillna(df_stress['installment'].median())
df_stress['grade'] = df_stress['grade'].fillna("Unknown")
```

```python
baseline_portfolio_pd = df_stress["defaultFlag"].mean()
baseline_portfolio_pd
```

```python
pd_by_grade = df_stress.groupby("grade")["defaultFlag"].mean()
df_stress['pd_base'] = df_stress["grade"].map(pd_by_grade).fillna(baseline_portfolio_pd)
df_stress[["grade", "pd_base"]].head()
```

```python
df_stress["pti_base"] = df_stress["installment"] / (df_stress["annual_inc"] / 12.0)

df_stress.loc[np.isinf(df_stress["pti_base"]), "pti_base"] = np.nan
df_stress["pti_base"] = df_stress["pti_base"].fillna(df_stress["pti_base"].median())
```

```python
df_stress["pti_base"].describe()
```

#### Helper függvények


```python
def clamp(x, lo=0.0, hi=1.0):
    return np.minimum(np.maximum(x, lo), hi)

def run_scenario_once(df_in: pd.DataFrame, scen: dict, alpha: float = 0.8, seed: int = 42) -> pd.DataFrame:
    rng = np.random.default_rng(seed)
    out = df_in.copy()
    n = len(out)

    inc_shock = rng.normal(scen["income_shock_mu"], scen["income_shock_sigma"], size=n)
    inc_mult = clamp(1.0 + inc_shock, lo=0.3, hi=1.2)

    unemp = rng.random(n) < scen["unemp_prob"]
    inc_mult = np.where(unemp, inc_mult * 0.5, inc_mult) 

    pay_mult = 1.0 + scen["payment_shock"]

    out["annual_inc_stress"] = out["annual_inc"] * inc_mult
    out["installment_stress"] = out["installment"] * pay_mult

    out["pti_stress"] = out["installment_stress"] / (out["annual_inc_stress"] / 12.0)
    out["pti_rel"] = out["pti_stress"] / out["pti_base"]

    out["pd_stress_raw"] = out["pd_base"] * (1.0 + alpha * (out["pti_rel"] - 1.0))
    out["pd_stress"] = out["pd_stress_raw"].clip(lower=0.0001, upper=0.80)

    return out
```

```python
summary = []

for name, scen in SCENARIOS.items():
    tmp = run_scenario_once(df_stress, scen, alpha=0.8, seed=123)
    stress_pd = tmp["pd_stress"].mean()

    summary.append({
        "scenario": name,
        "baseline_pd_observed": baseline_portfolio_pd,
        "stress_pd_mean": stress_pd,
        "uplift_vs_baseline": stress_pd / baseline_portfolio_pd,
        "target_uplift": scen["pd_uplift_target"]
    })

stress_summary = pd.DataFrame(summary)
stress_summary
```

```python
scen_name = "adverse"
tmp = run_scenario_once(df_stress, SCENARIOS[scen_name], alpha=0.8, seed=123)

grade_impact = (
    tmp.groupby("grade")
       .agg(
           n=("grade", "size"),
           pd_base=("pd_base", "mean"),
           pd_stress=("pd_stress", "mean"),
       )
)

grade_impact["uplift"] = grade_impact["pd_stress"] / grade_impact["pd_base"]
grade_impact.sort_values("uplift", ascending=False)
```

```python
plt.figure()

plt.bar(
    stress_summary["scenario"],
    stress_summary['stress_pd_mean']
)

plt.axhline(
    y=stress_summary['baseline_pd_observed'].iloc[0],
    linestyle="--",
    label="Baseline PD"
)

plt.ylabel("Portfólió PD")
plt.title("Portfólió PD stressz szcenáriók szerint")
plt.legend()
plt.show()
```

## Mit lehetne még javítani/fejleszteni?

Ebben a projektben az volt a célom, hogy építsek egy működő MVP-t, ezért a stressztesztnél egy egyszerűbb, logikusnak tűnő becslést használtam. Viszont miközben kutattam a téma után, láttam, hogy egy igazi banki környezetben (pl. IFRS 9 alatt) ennél azért szofisztikáltabb módszerek kellenek.

A jövőben, ahogy időm engedi (vagy élesben dolgoznék rajta), ezeket fejleszteném tovább:

### 1. "Kézi szorzó" helyett Logisztikus Regresszió
Most egy fix szorzóval (`alpha`) állítottam be, hogy mennyire rontsa a fizetési nehézség a csőd valószínűségét. Ez egy jó közelítés demo projektben, de a valóságban ez nem lineáris.
* A "Profi" megoldás: Építenék egy Logisztikus Regressziós modellt. Ez sokkal pontosabban megmondaná, hogy ha esik a jövedelem (GDP sokk), az pontosan mennyivel növeli a PD-t.
* Itt olvastam róla:
    * *Naeem Siddiqi: Intelligent Credit Scoring* – Azt láttam, hogy ez az alapkönyv a témában és MI segítségével kigyűjtöttem a lényeges részeket belőle, a későbbiekben elszeretném olvasni teljes egészében. [Link](https://www.wiley.com/en-us/Intelligent+Credit+Scoring%3A+Building+and+Implementing+Better+Credit+Risk+Scorecards%2C+2nd+Edition-p-9781119279150)
    * *Bank of England tanulmány* – Itt írják le, hogyan hatnak a makro sokkok a PD-re. [Link](https://www.bankofengland.co.uk/working-paper/2014/stress-testing-banks-a-model-to-assess-the-impact-of-macroeconomic-shocks-on-pds)

### 2. Éves átlag helyett "Vintage Görbék"
Most éves átlagokkal számoltam, ami egyszerű, de elrejti azt, hogy a hitelek kockázata változik az életciklusuk során (pl. a 12-18. hónapban dőlnek be a legtöbben).
* A "Profi" megoldás: Kirajzolnám a Vintage görbéket a folyósítás dátuma (`issue_d`) alapján. Így látnánk, hogy egy 2015-ös hitel másképp viselkedik-e, mint egy 2016-os.
* Forrás: Investopedia: Vintage Analysis – Itt értettem meg, miért fontos a "maturation". [Link](https://www.investopedia.com/terms/v/vintage_analysis.asp)

### 3. A 3 és 5 éves hitelek szétválasztása
Most egy kalap alá vettem a 36 és 60 hónapos futamidejű hiteleket.
* A "Profi" megoldás: Külön bontanám őket, mert logikus, hogy 5 év alatt több minden történhet az ügyféllel (bizonytalanság nagyobb), így a kockázatuk is más.
* Forrás: LendAcademy – Itt találtam adatokat arról, hogy a LendingClubnál is mennyire eltér a két típus teljesítménye. [Link](https://www.lendacademy.com/lending-club-loan-performance-vintage/)


Összességében: A jelenlegi modellem jó arra, hogy lássuk a trendeket, de tudom, hogy a szabályozói elvárásokhoz (és a pontos tőkeszámításhoz) a fenti lépésekkel kellene továbbfejleszteni és a jelenlegi állapot nem "production-ready".
